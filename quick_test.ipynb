```python
def get_grade(marks):
    grades = {
        90: "A",
        75: "B",
        60: "C",
        0: "D"
    }

    for threshold, grade in grades.items():
        if marks >= threshold:
            return grade
```

```python
def calculate_grade(name, marks):

    if marks >= 90:
        grade = "Grade A"

    elif marks >= 75:
        grade = "Grade B"

    elif marks >= 60:
        grade = "Grade C"

    else:
        grade = "Grade D"

    return name, marks,grade


for i in students:
    result = calculate_grade(i['name'], i['marks'])
    print(result)
```

```python
sales = [
    {"product": "Laptop", "price": 50000, "quantity": 2},
    {"product": "Mobile", "price": 20000, "quantity": 3},
    {"product": "Tablet", "price": 15000, "quantity": 4}
]


import pandas as pd


df = pd.DataFrame(sales)




grand_total  = 0
for i in sales:
    grand_total += i['price'] * i['quantity']

def calculate_revenue(price, quantity):
    revenue = price * quantity
    return revenue
for rev in sales:
    calling_fn = calculate_revenue(rev['price'],rev['quantity'])
    print(rev['product'],calling_fn)
print(f"Grand Total of all product is {grand_total}")


```

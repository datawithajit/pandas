```python
import csv

input_path = r"/home/ajit/pd/students copy.csv"
output_path = r"/home/ajit/pd/students_summary.csv"  # The new file path

with open(input_path, mode="r", encoding="UTF-8") as infile:
    rows = csv.reader(infile)
    
    # 1. Read the old header and append the two new column names
    header = next(rows)
    new_header = header + ["Total_Marks", "Percentage"]
    
    # Prepare a list to collect all our new row data
    updated_rows = []
    
    # 2. Process each student
    for row in rows:
        if not row:
            continue
            
        # Convert subject marks to integers (from index 2 to the end)
        marks = [int(mark) for mark in row[2:]]
        
        # Calculate totals safely without using the built-in sum() function
        total_marks = 0
        for m in marks:
            total_marks += m
            
        percentage = round(total_marks / len(marks), 2)
        
        # 3. Create the new row by adding our calculations to the original data
        # Example: ['101', 'Aarav', '85', '90', '78', '88', '95', 436, 87.2]
        new_row = row + [total_marks, percentage]
        updated_rows.append(new_row)

# 4. Open the new file and write all the updated data into it
with open(output_path, mode="w", encoding="UTF-8", newline="") as outfile:
    writer = csv.writer(outfile)
    
    # Write the new header first
    writer.writerow(new_header)
    
    # Write all student data rows at once
    writer.writerows(updated_rows)

print(f"Success! Saved updated data with new columns to: {output_path}")
```
